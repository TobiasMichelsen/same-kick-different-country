
# Google Trends – Exploratory Data Analysis

## Objective
Explore Google Trends as a secondary data source
for analyzing techno music across European countries.

Initial focus: Denmark
Time period: Past 12 months

Data sources:
- Interest over time
- Interest by region
- Top related queries
- Rising related queries


```python

import pandas as pd
from pathlib import Path

data_dir = Path("../data/raw/google_trends")

files = {
    "time_series": "time_series_DK.csv",
    "regions": "by_region_DK.csv",
    "top_queries": "top_queries_DK.csv",
    "rising_queries": "rising_queries_DK.csv"
}

for name, filename in files.items():
    df = pd.read_csv(data_dir / filename)
    print(f"\n{name}: {df.shape}")
    display(df.head())
```
