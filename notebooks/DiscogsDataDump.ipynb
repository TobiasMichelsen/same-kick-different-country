```python
## from Claude! ! !

import pandas as pd
from lxml import etree

# sample.xml is truncated mid-release, so use a recovering parser
parser = etree.XMLParser(recover=True, huge_tree=True)
root = etree.parse("/Users/tobias/sample.xml", parser).getroot()

def texts(el, path):
    """Join the text of all elements matching path."""
    return ", ".join(e.text for e in el.findall(path) if e.text)

rows = []
for r in root.iter("release"):
    rows.append({
        "id": int(r.get("id")),
        "status": r.get("status"),
        "title": r.findtext("title"),
        "artists": texts(r, "artists/artist/name"),
        "labels": ", ".join(l.get("name", "") for l in r.findall("labels/label")),
        "catno": ", ".join(l.get("catno", "") for l in r.findall("labels/label")),
        "formats": ", ".join(f.get("name", "") for f in r.findall("formats/format")),
        "genres": texts(r, "genres/genre"),
        "styles": texts(r, "styles/style"),
        "country": r.findtext("country"),
        "released": r.findtext("released"),
        "num_tracks": len(r.findall("tracklist/track")),
        "data_quality": r.findtext("data_quality"),
    })

df = pd.DataFrame(rows)
print(df.shape)
df.head()
```
